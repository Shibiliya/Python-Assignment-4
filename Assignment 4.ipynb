```python
def filter_strings(strings):
    result=[]
    for s in strings:
        if len(s) >=5:
            result.append(s)
    return result
```

```python
strings=["apple","dog","banana","cat","elephant"]
result=filter_strings(strings)
print(result)
```

```python
def is_prime(n):
    if n<=1:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True
numbers=[2,3,4,5,6,7,8,9,10]
primes=list(filter(is_prime,numbers))
print(primes)
```

```python
def to_uppercase(strings):
    result=[]
    for s in strings:
        result.append(s.upper())
    return result
strings=["apple","dog","banana","cat","elephant"]
uppercase_strings=to_uppercase(strings)
print(uppercase_strings)
```
